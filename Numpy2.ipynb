#### for an array [4,6,8,10,12]. calculate mean ,median ,std

```python
arr=np.array([4,6,8,10,12])
c=np.mean(arr)
print(c)
print(np.std(arr))
print(np.sort(arr))
print(np.median(arr))
```

### or better way 

```python
data=np.array([4,6,8,10,12])
mean_data=np.mean(data)
medain_data=np.median(data)
std_data=np.std(data)
print(mean_data)
print(medain_data)
print(std_data)
```
